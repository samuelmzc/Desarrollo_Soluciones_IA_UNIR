Como todos los lenguajes de programación, Python tiene variables, métodos, listas, bucles, entre otros. Vamos a hacer un repaso de lo más importante para el mundo de los datos.

Python permite almacenar distintos tipos de datos en *listas*. Se puede acceder a los elementos de la lista concatenando corchetes a la variable que la contiene, e indicando el índice del elemento que queremos encontrar (empieza por 0).

```python
lista = [0, 2, 3, "pepe", 2.1]
lista[1]
```

En la siguiente celda vemos algunos métodos y usos específicos que tienen las listas.

```python
lista.append(2) # Añade al final de la lista otro elemento, el int 2
lista.pop() # Elimina de la lista el último elemento
lista[-1] # Se pueden usar índices negativos, esto indica que se empieza la enumeración desde el final
lista.remove("pepe") # Elimina el elemento indicado
```

El lenguaje de por sí no es eficiente a la hora de aplicar operaciones aritméticas sobre estas listas. E.g. Vamos a intentar obtener el cuadrado de los elementos de una lista. Esto se hace mediante un bucle, de dos formas.

```python
lista = [1, 2, 3, 4, 1, 1, 2, 2, 3, 4, 1, 6, 7]

# Bucle for
lista_cuadrado1 = []
for i in lista:
    lista_cuadrado1.append(i**2) # Aquí, el método append lo que hace es añadir a la lista el elemento

# Lógica de listas: bucle dentro de una lista
lista_cuadrado2 = [i**2 for i in lista]

print(lista_cuadrado1)
print(lista_cuadrado2)
```

Cuando tratamos datos, es necesario encontrar una forma de optimizar estos cálculos, pues se van a hacer a gran escala. De aquí nace *NumPy* (**Num**erical **Py**thon), una librería que optimiza las operaciones hacia las listas, mediante *Arrays* (En general numpy ofrece un conjunto de métodos y funciones ampliamente aplicadas en el mundo de la ciencia y la ingeniería, más en https://numpy.org/)

Se puede importar una librería de dos formas distintas, con **import**, o con **from**. Con import, podremos usar los métodos, funciones, etc que numpy ofrece poniendo delante el nombre de la librería (e.g. array = numpy.array), pudiendo usar **as** para poner un alias (numpy de forma estándar se abrevia como np). Usando from, podemos importar funciones específicas, o todas ellas mediante *, de forma que no hay que especificar la librería a la hora de usarla.

```python
from numpy import *
array = array([1, 2, 3])
```

```python
import numpy as np
array = np.array([1, 2, 3])
```

Realizar operaciones sobre arrays es sencillo:

```python
array2 = array ** 2
array2
```

Se pueden aplicar lógicas a los arrays, esto va a ser muy útil a la hora de explorar los datos. Imaginémonos que del array de la siguiente celda, queremos que nos devuelva ese mismo array pero con los elementos menores a 4.

```python
a = np.array([1, 2, 3, 1, 1, 5, 6])
```

Primero veamos la siguiente celda.

```python
a < 4
```

Esta lógica me está devolviendo un array con booleanos, cuyo valor será True si se cumple la condición (el número es menor que 4), y False en caso contrario. Basta con poner esta lógica entre corchetes para obtener el array que necesitamos.

**Nota**: Fíjate que el tamaño (shape) es menor que el original.

```python
a[a < 4]
```

También se pueden concatenar estas lógicas mediante los operadores lógicos AND (and, &), y OR (or, |)

```python
a[(a < 4) | (a == 5)]
```
