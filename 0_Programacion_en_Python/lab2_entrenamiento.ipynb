```python
import numpy as np
import pandas as pd

# Nos dan la siguiente lista con productos
ventas = [
    {"Producto": "Laptop", "Categoría": "Electrónica", "Unidades": 5, "Precio_unitario": 800},
    {"Producto": "Smartphone", "Categoría": "Electrónica", "Unidades": 10, "Precio_unitario": 600},
    {"Producto": "Impresora", "Categoría": "Oficina", "Unidades": 7, "Precio_unitario": 150},
    {"Producto": "Monitor", "Categoría": "Electrónica", "Unidades": 3, "Precio_unitario": 200},
    {"Producto": "Teclado", "Categoría": "Oficina", "Unidades": 15, "Precio_unitario": 25},
    {"Producto": "Smartphone", "Categoría": "Electrónica", "Unidades": 4, "Precio_unitario": 600},
    {"Producto": "Laptop", "Categoría": "Electrónica", "Unidades": 2, "Precio_unitario": 800},
    {"Producto": "Teclado", "Categoría": "Oficina", "Unidades": 10, "Precio_unitario": 25}
]

# Primero creamos el DataFrame
df_ventas = pd.DataFrame(ventas)

# Mostramos primero las 5 primeras filas, y luego las 5 últimas
print("Primeras 5 columnas")
print(df_ventas.head())

print("\nÚltimas 5 columnas")
print(df_ventas.tail())

# Resumen estadístico
print("\nResumen estadístico")
print(df_ventas.describe())

# Cantidad de productos vendidos por categoría
print("\nCantidad de productos vendidos por categoría")
print(df_ventas["Categoría"].value_counts())

# Estructura del DataFrame
print("\nEstructura del DataFrame")
df_ventas.info()

# Nueva columna con precio total
df_ventas["Total"] = df_ventas.Unidades * df_ventas.Precio_unitario
print("\nAñadimos nueva columna: Total")
print(df_ventas)

# Añadimos 10% de descuento a productos de categoría oficina
df_ventas["Precio_unitario"] = df_ventas["Precio_unitario"].astype(float) #Convertimos el precio unitario en float
df_ventas.loc[df_ventas.Categoría == "Oficina", "Precio_unitario"] *= 0.9
df_ventas["Total"] = df_ventas.Unidades * df_ventas.Precio_unitario
print("\n Aplicamos 10% descuento a productos de categoría oficina")
print(df_ventas)

# Seleccionamos los productos de categoría Electrónica con más de 5 unidades vendidas
selection = df_ventas.loc[(df_ventas.Categoría == "Electrónica") & (df_ventas.Unidades > 5)]
print("\nProductos de Electrónica con más de 5 unidades vendidas")
print(selection)
```

```python
import numpy as np
import pandas as pd

# Cargamos los datos
inventarios = pd.read_csv("../media/0_lab2/inventarios.csv")
satisfaccion = pd.read_csv("../media/0_lab2/satisfaccion.csv")
ventas = pd.read_csv("../media/0_lab2/ventas.csv")

# Se limpian los valores nulos
inventarios.dropna(inplace=True)
satisfaccion.dropna(inplace=True)
ventas.dropna(inplace=True)

# Veamos información de las tablas
inventarios.info()
satisfaccion.info()
ventas.info()

# Mostramos el dataframe de ventas
print("DataFrame de ventas")
print(ventas)

# Ingresos totales por tienda
print("\nIngresos totales por tiendas")
ventas["Ingresos_Totales"] = ventas.Cantidad_Vendida * ventas.Precio_Unitario
ventas_por_tienda = ventas.groupby("ID_Tienda")["Ingresos_Totales"].sum() 
print(ventas_por_tienda)

# Resumen estadístico
print("\nResumen estadístico de las ventas")
print(ventas.describe())

# Promedio de ventas por tienda y categoría
print("\nPromedio de ventas por tienda")
ventas_prom_por_tienda = ventas.groupby("ID_Tienda")["Cantidad_Vendida"].mean()
print(ventas_prom_por_tienda)

print("\nPromedio de ventas por categoría")
ventas_prom_por_prod = ventas.groupby("Producto")["Cantidad_Vendida"].mean()
print(ventas_prom_por_prod)

# Pasamos a inventarios. Calculamos la rotación de inventarios por tienda
inventarios["Rotación"] = ventas.Cantidad_Vendida / inventarios.Stock_Disponible
print("\nAñadimos la columna de Rotación")
print(inventarios)
print("\nRotación de inventarios por tienda")
rotacion_por_tienda = inventarios.groupby("ID_Tienda")["Rotación"].sum() 
print(rotacion_por_tienda)

# Veamos las tiendas en niveles críticos
cantidad_por_tienda = ventas.groupby("ID_Tienda")["Cantidad_Vendida"].sum()
stock_por_tienda = inventarios.groupby("ID_Tienda")["Stock_Disponible"].sum()
porcentaje_por_tienda = cantidad_por_tienda / stock_por_tienda
tiendas_criticos = cantidad_por_tienda.loc[porcentaje_por_tienda < 0.1]
print("\nTiendas en crítico")
print(tiendas_criticos)
print("No hay ninguna tienda con niveles críticos de inventario")

# Ahora a la satisfacción
print(satisfaccion)
print(rotacion_por_tienda)
filtro_satisfaccion = satisfaccion.loc[satisfaccion.Satisfacción_Promedio < 60]
filtro_satisfaccion
print("Un análisis de los datos nos hace comprobar que a mayor rotación, mayor satisfacción, así que podría probar a aumentar la rotación")

# Numpy
mediana_ventas_totales = np.median(ventas_por_tienda.to_numpy())
std = np.std(ventas_por_tienda.to_numpy())

# Asumiendo que se distribuye de forma normal, mediana = promedio
np.random.seed(222100)
simulacion_futura = np.random.normal(loc = mediana_ventas_totales, scale = std, size = len(ventas_por_tienda))
print(simulacion_futura)
simulaciones = pd.DataFrame({"ID_Tienda" : np.array([1, 2, 3, 4, 5]), "Simulación" : simulacion_futura})
print("Veamos estadísticas de las simulaciones")
print(simulaciones.describe())
```
